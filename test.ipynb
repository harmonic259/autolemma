```python
from autolemma.core import Var, And, Or, Not, tseitin_cnf, Const, Implies, Iff
from autolemma.algos.sat import bruteforce_solve, sat2_solve, dpll_solve, dpll_plus_solve
```

### Example 2.1

```python
# forrmula f1 = (P implies Q) implies (P and not R)
P = Var("P")
Q = Var("Q")
R = Var("R")
f1 = Implies(Implies(P, Q), And(P, Not(R)))
print("Formula f1:", f1)
print("NNF of f1:", f1.nnf())
print("Positions in f1:", sorted(f1.positions()))
print("Subformulas of f1:")
for pos in sorted(f1.positions()):
    print(" ", pos, ":", f1.subformula_at(pos))
print("Size of f1:", f1.formula_size())

# replace subformula at position 2.2. with (P iff Q)
f1_replaced = f1.replace_at_pos([2, 2], Iff(P, Q))
print("f1 after replacing subformula at position 2.2. with (P iff Q):", f1_replaced)
```
